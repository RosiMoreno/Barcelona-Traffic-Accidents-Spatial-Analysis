### 3. RESULTADOS: Presentación de los resultados obtenidos

#### **ANÁLISIS ESPACIAL**

![Analisis_Espacial_Accidentes_Barcelona](imagenes/Analisis_Espacial_Accidentes_Barcelona_2.png)

Las visualizaciones con QGIS según el distrito junto con el gráfico de cintas de PowerBI, nos muestran claramente que la zona más tensionada a nivel de distrito municipal es el Eixample. Presenta una gran diferencia respecto al distrito en segunda posición que es Sant Martí. Barrio que también ocupa una gran extensión de la planificación urbanística diseñada por Ildefonso Cerdà. Si visualizamos el mapa de calor de mayor escala se observa una gran mancha de color roja en el centro de la ciudad, construido en base a este plan urbanístico. La suma de los porcentajes de los cuatro distritos con más accidentes representa casi el 64% del total de accidentes ocurridos en Barcelona durante los años de estudio. 

El plan urbanístico tiene como base una distribución de manzanas con una essquinas recortadas en chaflán, entre las cuáles discurren grandes calles perpendiculares con 3 y 4 carriles de coches por calle. Dos de estas calles que atraviesan toda la ciudad son de mayor amplitud; la calle Aragó y Gran Via de les Corts Catalanas.  Además, hay dos avenidas que atraviesan la ciudad en la bisectriz del ángulo de 90 grados que separa a estas calles: La Diagonal y la Meridiana. Gran parte del tráfico de la ciudad se concentra por estas calles de mayor tamaño. Posteriormente, para liberar el tráfico de la ciudad, se crean varias rondas exteriores a la ciudad ( Ronda Litoral, Ronda de Dalt y Ronda del Mig). 

Si nos acercamos con un poco de zoom en el mapa de calor de Barcelona, se puede observar cómo la mayoría de los accidentes se distribuyen por estas grandes avenidas por las que circula más tráfico. Y vemos con mayor claridad que el gran foco de accidentes se centra en el Distrito del Eixample, con principal incidencia en Passeig de Gracia. 

![Mapa_Calor_Avenidas_Rondas_Accidentes_Barcelona](imagenes/Mapa_Calor_Avenidas_Rondas.png)

Al acercarnos un poco más en los alrededores de Passeig de Gracia en este mapa de calor, que esta zona de Barcelona pertenece al barrio de la Dreta de l'Eixample que es el barrio con más accidentes a lo largo del histórico de datos analizados en este estudio y hace referencia a casi la cuarta parte del total de accidentes de Barcelona (24,7%). También, se puede ver que el segundo barrio con más accidentes es l'Antigua Esquerra de l'Eixample con varios focos altos de accidentes. Esta informmación también nos la muestra el gráfico de barras realizadod en PowerBI. Este ranking de los dos primeros barrios con mayores accidentes se mantiene a lo largo de los años, únicamente va cambiando la tercera posición.

### 4. CONCLUSIONES: Principales conclusiones de los resultados obtenidos.

Una vez concluido el estudio de los datos de accidentes de Barcelona se puede decir que hay factores espaciales que determinan el riesgo de accidentes como son los cruces de las calles en el Eixample y también en otros puntos de la ciudad. También hay un factor de ubicación muy importante en la incidencia de accidentes puesto que casi el 64% de los accidentes de Barcelona se regitran en el distrito del Eixample. Esto nos da una pista a lo hora de tomar decisiones respecto a los equipamientos involucrados en un accidente. Es decir, habría que poner especial atención y cubrir suficientemente las instalaciones de este distrito. 

En ese aspecto, habría un interesante nicho de estudio respecto a las causas de que se produzcan los accidentes en estos puntos concretos y un estudio de prevención de estos accidentes, principalmente a nivel señalización y también concienciación de la importancia de esperar el turno de paso en los cruces. Por otro lado, ayudaría tener información de que tipo de vehículos han estado involucrados en los accidentes. Esto sería una información relevante a incluir en la base de datos. 

Desde el punto de vista temporal, hay un factor importante que determina el número de accidentes y es la hora, como se ha visto durante la tarde se producen en torno al 50% de los accidentes. Es verdad que la variable días de la semana, no es determinista, exceptuando que se puede observar una caída del 50% de accidentes durante los domingos. Esto nos ayuda a planificar al personal que necesitamos tener disponible en lso servicios implicados en la gestión de accidentes (servicio médico, bomberos, policias y teléfono de emergencia) y aumentarlo durante las horas más críticas y se puede reducir durante el domingo. Respecto a los meses, hay una reducción considerable durante los períodos vacacionales también a tener en cuenta en la decisión de la organización de los servicios.

Respecto a la evolución temporal a nivel global, se ha observado un evidente salto en número de accidentes después de la pandemia. Y durante el periodo posterior se ha mantenido un porcentaje importante por debajo del periodo pre-pandemia, pero a su vez presenta un ligero aumento. Así que la previsión de futuro, podría ser de aumento. Sin embargo, habría que hacer un estudio más detallado con modelos de análisis de tendencias y predicción más potentes.
